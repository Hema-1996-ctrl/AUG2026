# filtering (similar to where clause in SQL)

```python
# get me emp who are getting more than 3000 
# select *from emp where salary > 3000
salary_3000 = df['salary'] > 3000
```

```python
# we are getting only true or false 
df.salary_3000
```

```python
# to see all the data 
df[salary_3000]
```

```python
# get me emp who are getting more than 2000 and belongs to dept no 20
# select *from emp where salary > 3000 and deptno = 20
salary_2000 = df['salary'] > 2000 
deptno_20 = df['deptno'] == 20
final_filter = salary_2000 & deptno_20
```

```python
final_filter
```

```python
df[final_filter]
```

# sorting (ascending /descending)

```python
# sort based on salary column by asc
df.sort_values (by = 'salary')
# or
df.sort_values(by = 'salary', ascending = True)
```

```python
# sort based on salary column by desc
df.sort_values (by = 'salary')
# or
df.sort_values(by = 'salary', ascending = False)
```

```python
# sort based on deptno in ascending and salary  by desc
df.sort_values(by =['deptno','salary'], ascending = [True,False])
```
